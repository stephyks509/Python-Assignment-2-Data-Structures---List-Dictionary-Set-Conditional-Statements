a. Append the string "Yazhini" to name_list.

```python
name_list.append("Yazhini")
print("name_list:",name_list)
```

Explanation: append() adds one item to the end

b. Insert the element 30 at index 2 in age_list.

```python
age_list.insert(2,30)
print("age_list:",age_list)
```

Explanation: insert(i,x) inserts x at index i.(will insert multiple times if its runned)

d. Pop the last element from age_list.

```python
age_list.pop()
print(age_list)
```

```python
age_list.extend([29,30,26])
print(age_list)
```

Explanation: extend() adds every element of another list.

f. Sort age_list in descending order.

```python
age_list.sort(reverse=True)
print(age_list)
```

Explanation: sort(reverse=True) sorts the elements in the list from the largest to the smallest.

g. Find Max age, Min age and sum of all ages from age_list.

```python
print("Max age:",max(age_list))
print("Min age:",min(age_list))
print("Sum of all ages:",sum(age_list))
```

Explanation: The built-in functions max(), min() and sum() work on any iterable of numbers.

Written using { }

len() gives the no:of unique elements in the set.

set1 with values: {1, 3, 5, 7, 9}
set2 with values: {2, 3, 5, 8, 10}

d. Compute and print the union and intersection of set1 and set2.

```python
print(set1.union(set2))
```

```python
print(set1.intersection(set2))
```

Explanation:

Union returns all unique elements from both sets.

Syntax:

A | B

A.union(B)

Intersection returns only elements present in both sets.

Syntax:

A & B

A.intersection(B)
